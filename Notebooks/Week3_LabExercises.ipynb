```python
# Points class

class Point :
    """
    Represents a point in a 2D space
    
    attributes: x, y, name
    
    """
    
    # constructor function
    # The double underscores indicate a private method or variable
    # not to be accessed outside the class (in principle)
    def __init__(self, x=0.0, y=0.0, name = ""):
        self.x = x
        self.y = y
        self.name = name
        self.__private_variable = 42
        
    def __add__(self, other) :
        new_point = Point(self.x+other.x, self.y+other.y)
        return new_point
    
    def print_point(self) :
        print("Point ", self.name, "is", self.x, self.y)

    def reflect_point_x_axis(self):
        self.y = -self.y

    def reflect_point_y_axis(self):
        self.x = -self.x
        
    # Note that we don't use self here so don't need to pass it in
    # (This is a static function - it does not require an instance of the class)
    def calculate_distance_between_two_points(A, B) :
        return np.sqrt((A.x - B.x)**2.0 + (A.y - B.y)**2.0)
    
    def move_point(self,dx,dy) :
        self.x += dx
        self.y += dy
        
    def plot_point(self, ax) :
        ax.plot(self.x, self.y, 'o', label=self.name)
        
    def update_name(self, new_name) :
        self.name = new_name
        
        
# Some examples of using the Points class
first_point = Point(1.0,2.0,"P")
first_point.print_point()

second_point = Point(3.0,5.5,"Q")
second_point.move_point(1.0, 2.0)
second_point.print_point()

third_point = first_point + second_point
third_point.update_name("R")
third_point.print_point()

# Use the static function
distance = Point.calculate_distance_between_two_points(first_point,second_point)
print("Distance is ", distance)

plt.plot()
plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")
ax = plt.gcf().gca()
first_point.plot_point(ax)
second_point.plot_point(ax)
third_point.plot_point(ax)
plt.legend()

# Try to access the private variable, even though you shouldn't
#print(first_point.__private_variable) # This won't work - try uncommenting it
print(first_point._Point__private_variable) # This will

# Can get info on the new class - uncomment to see this
# Point?
```

```python
class Rectangle:
    def __init__(self, corner1, corner2):
        assert (type(corner1) == Point) and (type(corner2) == Point), "The corners must be Point types"
        assert not ((corner1.x == corner2.x) or (corner1.y == corner2.y)), "The provided corners should be opposite corners"
        self.corners = [corner1, corner2, Point(corner1.x, corner2.y), Point(corner2.x, corner1.y)]

    def _get_bottom_left_corner(self):
        x_coords = [corner.x for corner in self.corners]
        y_coords = [corner.y for corner in self.corners]
        return Point(min(x_coords), min(y_coords))

    def _get_top_right_corner(self):
        x_coords = [corner.x for corner in self.corners]
        y_coords = [corner.y for corner in self.corners]
        return Point(max(x_coords), max(y_coords))

    def reflect_x_axis(self):
        for corner in self.corners:
            corner.reflect_point_x_axis()

    def reflect_y_axis(self):
        for corner in self.corners:
            corner.reflect_point_y_axis()

    def scale_from_bottom_left(self, scale_factor):
        assert type(scale_factor) in [float, int] , "Scale factor must be a float or int"
        bottom_left = self._get_bottom_left_corner()

        for corner in self.corners:
            if corner.x == bottom_left.x and corner.y == bottom_left.y:
                continue
            else:
                corner.x = bottom_left.x + (corner.x - bottom_left.x) * scale_factor
                corner.y = bottom_left.y + (corner.y - bottom_left.y) * scale_factor
                
    def scale_from_top_right(self, scale_factor):
        assert type(scale_factor) in [float, int] , "Scale factor must be a float or int"
        top_right = self._get_top_right_corner()

        for corner in self.corners:
            if corner.x == top_right.x and corner.y == top_right.y:
                continue
            else:
                corner.x = top_right.x + (corner.x - top_right.x) * scale_factor
                corner.y = top_right.y + (corner.y - top_right.y) * scale_factor

    def plot_rectangle(self, ax):
        assert type(ax) == plt.Axes, "ax must be a matplotlib Axes object"
        for corner in self.corners:
            corner.plot_point(ax)
        #plot edges
        ax.plot([self.corners[0].x, self.corners[2].x], [self.corners[0].y, self.corners[2].y], 'b-')
        ax.plot([self.corners[2].x, self.corners[1].x], [self.corners[2].y, self.corners[1].y], 'b-')
        ax.plot([self.corners[1].x, self.corners[3].x], [self.corners[1].y, self.corners[3].y], 'b-')
        ax.plot([self.corners[3].x, self.corners[0].x], [self.corners[3].y, self.corners[0].y], 'b-')   
```

```python
point_1 = Point(1,3)
point_2 = Point(2.5, 2)

rect = Rectangle(point_1, point_2)


plt.plot()
plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")
ax2 = plt.gcf().gca()
rect.plot_rectangle(ax2)
rect.scale_from_top_right(2)
rect.plot_rectangle(ax2)
rect.reflect_x_axis()
rect.plot_rectangle(ax2)
```

Now create a Square class that has the same functionality as the rectangle class, but that in addition has the ability to calculate its own area. You are not allowed to modify the Rectangle class.

```python
# Square class

class Square(Rectangle):
    def __init__(self, bottom_left_corner, side_length):
        assert type(bottom_left_corner) == Point, "The bottom left corner must be a Point type"
        assert side_length > 0, "Side length must be positive"
        self.side_length = side_length
        top_right_corner = Point(bottom_left_corner.x + side_length, bottom_left_corner.y + side_length)
        super().__init__(bottom_left_corner, top_right_corner)

    def get_area(self):
        return self.side_length ** 2
    
```

```python
square = Square(Point(0, 0), 2)

plt.plot()
plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")
ax3 = plt.gcf().gca()
square.plot_rectangle(ax3)
square.get_area()
```

### ACTIVITY 3:

We now consider a simple system of coupled ODEs. The Lotka–Volterra equations describe predator–prey interactions between a prey population $x$ (for example, rabbits) and a predator population $y$ (for example, foxes). In the absence of predators, the prey population grows at a rate proportional to its size. In the absence of prey, the predator population declines at a rate proportional to its size. Predation occurs at a rate proportional to the product $xy$, reducing the prey population while supporting predator reproduction. The populations are therefore described by the following first-order, autonomous, nonlinear, coupled system of ordinary differential equations:

$\frac{dx}{dt} = ax - bxy$

$\frac{dy}{dt} = -cy + dxy$

Here, $a$ is the prey growth rate, $b$ measures predation, $c$ is the predator death rate, and $d$ measures how effectively consumed prey contributes to predator growth.

What parts or features of the differential equation tell us that it is:
1. First order
2. Autonomous
3. Non linear
4. Coupled 

?

Answers:

1. There is only a first derivative
2. There is no $t$ component to the equations, only $dt$
3. $xy$ tells us that these are non-linear as it is a product of variables
4. One describes the derivative of $x$ and one describes the derivative of $y$, but the functions that define these derivatives contain each other in there definitions

### ACTIVITY 4:

The following code solves the Logistic Equation 

$\frac{dy}{dt} = y (1 - y)$

Using a simple explicit integration and the scipy function solve_ivp() which used Runge-Kutta 4th/5th order time integration. Copy it and update it to solve instead the Lotka-Volterra equations. 

Note the following:

1. Why do I substract delta_t when calculating the vector of t_solution points?
2. the solve_ivp method uses the labels t and y for the independent and dependent variables. The y value is assumed to be a vector, allowing you to solve for multiple dependent variables at once, e.g. you can make it np.array([x,y])

```python
# Solve the 1d logistic equation from class
from scipy.integrate import solve_ivp

# Note that the function has to take t as the first argument and y as the second
def calculate_logistic_dydt(t, y):
    """Returns the gradient dx/dt for the logistic equation"""
    dydt = y*(1 - y)
    return dydt

max_time = 10.0
N_time_steps = 25
delta_t = max_time / N_time_steps
t_solution = np.linspace(0.0, max_time-delta_t, N_time_steps) # values of independent variable 
y0 = np.array([0.5]) # an initial condition, y(0) = y0, note it needs to be an array

solution = solve_ivp(calculate_logistic_dydt, [0,max_time], y0,
                        method='RK45', t_eval=t_solution)
 
plt.grid()
plt.xlabel("t", fontsize=16)
plt.ylabel("y", fontsize=16) 
plt.plot(solution.t, solution.y[0], 'o', label="solve_ivp")

# Now do it the "cheap" way 
# increase the number of steps to see how the solution changes
y_solution = np.zeros_like(t_solution)
y_solution[0] = y0[0]
for itime, time in enumerate(t_solution) :
    if itime > 0 :
        dydt = calculate_logistic_dydt(time, y_solution[itime-1])
        y_solution[itime] = y_solution[itime-1] + dydt * delta_t
    
plt.plot(t_solution, y_solution, '-.',label="cheap method")

# Now plot the true solution
A = 1.0/y0 - 1.0
y_true = 1.0 / (1.0 + A * np.exp(-t_solution))
plt.plot(t_solution, y_true, '-', label="true solution")
plt.legend(loc='best');
```

```python
# Implementation of the Ecosystem class (OPTIONAL)

class Ecosystem:
    def __init__(self, prey_population, predator_population, a, b, c, d):
        self.prey_population = prey_population
        self.predator_population = predator_population
        self.a = a  # Prey growth rate
        self.b = b  # Prey death rate due to predation
        self.c = c  # Predator death rate
        self.d = d  # Predator growth rate due to predation

    def calculate_dydt(self, t, y):
        prey = y[0]
        predator = y[1]

        dydt = np.zeros_like(y)
        dydt[0] = self.a * prey - self.b * prey * predator  # Prey equation
        dydt[1] = self.d * prey * predator - self.c * predator # Predator equation

        return dydt

    def simulate(self, max_time, N_time_steps):
        delta_t = max_time / N_time_steps
        t_solution = np.linspace(0.0, max_time-delta_t, N_time_steps) # values of independent variable 
        y0 = np.array([self.prey_population, self.predator_population]) # initial conditions

        solution = solve_ivp(self.calculate_dydt, [0,max_time], y0,
                             method='RK45', t_eval=t_solution)
        
        return solution
```

```python
forest = Ecosystem(prey_population=1, predator_population=0.1, a=0.5, b=0.2, c=0.2, d=0.1)

solution = forest.simulate(max_time=100.0, N_time_steps=2500)

plt.grid()
plt.xlabel("t", fontsize=16)
plt.ylabel("y", fontsize=16) 
plt.plot(solution.t, solution.y[0], 'o', label="Rabbit population")
plt.plot(solution.t, solution.y[1], 'o', label="Fox population")
plt.legend()
```
